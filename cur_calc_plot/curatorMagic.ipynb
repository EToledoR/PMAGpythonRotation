```python
# Settings for data cleaning

# Minimum number of samples required. Set to -1 to ignore.
num_min_samples = -1

# Threshold for removing overrepresented locations.
location_removal_threshold = 100000

# Estimate missing A95 and K values where possible.
estimate_missing_a95_k = True

# Fill missing N values using N_alt where possible.
fill_sites = True

# Use a custom column mapping instead of the default mapping.
custom_map_needed = True

custom_map = {
    "nSample": "N",
    "Nsites": "N_alt",
    "Slat": "slat",
    "Slon": "slon",
    "k": "k",
    "a95": "a95",
    "Dec": "mdec",
    "Inc": "minc",
    "Geologic Area": "loc",
    "AgeLowBound": "AgeLowBound",
    "AgeUpBound": "AgeUpBound",
    "Name": "name",
    "Ref": "Ref",
    "A95_VGP": "A95",
    "K_VGP": "K",
}

```

```python
### CURATE THE DATA
from pathlib import Path

cwd = Path.cwd()

REPO_ROOT = next(
    (
        parent
        for parent in (cwd, *cwd.parents)
        if (
            parent
            / "data"
            / "CustomTestData.csv"
        ).is_file()
    ),
    None,
)

if REPO_ROOT is None:
    raise FileNotFoundError(
        "Could not locate the repository root."
    )

data_path = REPO_ROOT / "data" / "CustomTestData.csv"

if not data_path.is_file():
    raise FileNotFoundError(f"Input dataset not found: {data_path}")

print(f"Repository root: {REPO_ROOT}")
print(f"Input dataset: {data_path}")

df = import_data(str(data_path))

# Add required metadata columns
df["Primary"] = "Y"
df["Collection"] = ""

# Run the curation workflow
tec_data = curator(
    df,
    num_min_samples,
    location_removal_threshold,
    estimate_missing_a95_k,
    fill_sites,
    custom_map_needed,
    custom_map,
)

# Export curated data
output_dir = REPO_ROOT / "data"
f1 = output_dir / "tec_c.csv"
f2 = output_dir / "tec_c2.csv"

pole_and_rotation_export(
    tec_data,
    str(f1),
    str(f2),
)


```
